```python
import pandas as pd
import os
import logging

logging.basicConfig(level=logging.INFO, format='%(asctime)s | %(levelname)-7 | %(message)s')

def run_transformation_and_tests():
    logging.info("=== Tech Catalog: Transformation Rules & Quality Tests Started ===")
    
    input_dir = "."
    
    try:
        ksa_extra = pd.read_csv(os.path.join(input_dir, "cleaned_KSA_Extra_Products.csv"))
        ksa_google = pd.read_csv(os.path.join(input_dir, "cleaned_KSA_Google_Devices.csv"))
        us_extra = pd.read_csv(os.path.join(input_dir, "cleaned_US_Extra_Products.csv"))
        us_google = pd.read_csv(os.path.join(input_dir, "cleaned_US_Google_Devices.csv"))
    except Exception as e:
        logging.error(f"Error loading source files: {e}")
        return

  
    def clean_cols(df):
        df.columns = [str(col).strip().lower() for col in df.columns]
        return df

    ksa_extra, ksa_google = clean_cols(ksa_extra), clean_cols(ksa_google)
    us_extra, us_google = clean_cols(us_extra), clean_cols(us_google)

    ksa_extra['market'] = 'KSA'
    ksa_google['market'] = 'KSA'
    us_extra['market'] = 'US'
    us_google['market'] = 'US'

    final_ksa = pd.concat([ksa_extra, ksa_google], ignore_index=True)
    final_us = pd.concat([us_extra, us_google], ignore_index=True)
    logging.info("Transformation rules applied: Sources merged and markets tagged.")

    if 'product_price' in final_ksa.columns and 'product_title' in final_ksa.columns:
        issues_ksa = final_ksa[final_ksa['product_price'].isna() | final_ksa['product_title'].isna()].shape[0]
        logging.info(f"[KSA] Quality test completed: {issues_ksa} issue(s) flagged (missing prices or titles).")
    else:
        logging.warning("[KSA] Expected columns not found.")

    if 'product_price' in final_us.columns and 'product_title' in final_us.columns:
        issues_us = final_us[final_us['product_price'].isna() | final_us['product_title'].isna()].shape[0]
        logging.info(f"[US] Quality test completed: {issues_us} issue(s) flagged (missing prices or titles).")
    else:
        logging.warning("[US] Expected columns not found.")

    
    final_ksa.to_csv("final_KSA_products.csv", index=False)
    final_us.to_csv("final_US_products.csv", index=False)
    logging.info(f"Successfully exported final KSA ({len(final_ksa)} rows) and US ({len(final_us)} rows) datasets.")
    
    logging.info("=== Pipeline Completed Successfully ===")

if __name__ == "__main__":
    run_transformation_and_tests()
```
